# Your First Claude Agent SDK Call

The Claude Agent SDK gives you the same agent loop that powers Claude Code, packaged as a Python library you can call from your own code. `query()` is the simplest entry point: give it a prompt, get back a stream of messages as the agent works through it.


## Setup

Install inside the project's conda env (already done for this notebook's kernel):

```bash
conda activate .claude-agent-sdk-youtube-env
pip install claude-agent-sdk
```


## Anthropic API key

How to add your Anthropic API key:

1. Get a key from the Anthropic Console: https://console.anthropic.com/settings/keys
2. Create a `.env` file in the project root (it's git-ignored, so it stays local).
3. Add this line to it:
    ```
    ANTHROPIC_API_KEY=your-key-here
    ```
4. Load it with a package like `python-dotenv`, or export it in your shell before
   launching the notebook/script:
    ```bash
    export ANTHROPIC_API_KEY=your-key-here
    ```
5. In code, only ever read it via `os.environ["ANTHROPIC_API_KEY"]` — never open
   or print the `.env` file itself.

The Claude Agent SDK picks up `ANTHROPIC_API_KEY` from the environment
automatically, so as long as it's set before `query()` runs, no extra
configuration is needed here.


````python
# query() is the main function of the Claude Agent SDK.
# You give it a prompt (a question or task), and it runs Claude in an agent loop
# and streams back everything that happens, one message at a time.
from claude_agent_sdk import query


async def ask_claude(prompt: str) -> None:
    # prompt: the text you want to send to Claude, e.g. "What is 2 + 2?"

    # query() doesn't return one answer — it returns a stream (async generator)
    # of messages. We loop over that stream with "async for" and print each one
    # as it arrives, instead of waiting for everything to finish first.
    async for message in query(prompt=prompt):
        # Each `message` is one step of Claude's work. Common types you'll see:
        #   SystemMessage    -> setup info (session id, model, tools available, etc.)
        #   RateLimitEvent   -> info about your current usage limits
        #   AssistantMessage -> Claude's actual reply (the text you care about)
        #   ResultMessage    -> the final summary (cost, duration, success/failure)
        print(message)
````

````python
# We're inside a Jupyter notebook, so we can just "await" the async function directly.
# Outside a notebook (a plain .py script), you'd instead write:
#     import asyncio
#     asyncio.run(ask_claude("What is 2 + 2?"))
await ask_claude(
    "There is a Family, and in which 7 brother has 1 sister each, then how many member in the family we have? Give me the final count, and dont give me the explaination, only give me the final count number"
)
````

## Summary

- `query()` starts the same agent loop that powers Claude Code and streams back messages as they happen.
- This is the simplest possible use of the SDK — no options, no tools, no hooks. Later episodes build on this one call.
